```python

reordered_dir =  "/path/to/CEBPB_seq/ChromHMM/All_cellTypes_histones/numstates_22_reordered/"
emissions_fpath = Path(reordered_dir) / "emissions_22.txt"
seg_fpath_out= Path(reordered_dir) / "Histones_22_segments.bed"

emissions_out_pdf= Path(reordered_dir) / "Histones_STACKED_ChromHMM_22_reordered.pdf"
```

```python

marks=[
'H3K9ac', 
'H3K4me3', 
'H3K4me2', 
'H3K4me1', 
'H3K27ac',
'H3K27me3', 
'H3K79me2',
'H3K36me3', 
'H3K9me3']


cell_types=["MEF",
          "NSC",
            "ESC",
            "BMpreB"
           ]
col_order=[f"{cell}_{mark}"  for mark in marks for cell in cell_types]
col_order
```

```python
import pandas as pd

# load
emissions_df = pd.read_csv(emissions_fpath, sep="\t", index_col=0)
# state_col = "State (Emission order)"

# reorder columns
emissions_df = emissions_df[col_order]

em_mat = emissions_df.copy()
em_mat
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from matplotlib.colors import LinearSegmentedColormap
import matplotlib as mpl

# --- Global style (safe for publications)
import matplotlib as mpl
mpl.rcParams.update({
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "pdf.use14corefonts": True,   # use standard PDF fonts (Helvetica etc.)
    "font.family": "sans-serif",
    "font.sans-serif": ["Helvetica", "Arial", "DejaVu Sans"],
})

# Ensure values are percent (ChromHMM style)
# If your values are 0–1:
em_plot = em_mat.copy() * 100   # remove *100 if already 0–100

# ChromHMM white → blue colormap
chromhmm_cmap = LinearSegmentedColormap.from_list(
    "chromhmm_white_blue",
    ["#FFFFFF", "#DEEBF7", "#9ECAE1", "#4292C6", "#08519C"]
)

chromhmm_white_orange = LinearSegmentedColormap.from_list(
    "chromhmm_white_orange",
    ["#FFFFFF", "#FEE6CE", "#FDBE85", "#FD8D3C", "#D94701"]
)

chromhmm_white_purple = LinearSegmentedColormap.from_list(
    "chromhmm_white_purple",
    ["#FFFFFF", "#EFEDF5", "#BCBDDC", "#756BB1", "#54278F"]
)

# chromhmm_cmap = LinearSegmentedColormap.from_list(
#     "chromhmm_soft",
#     ["#FFFFFF", "#F0F6FB", "#C6DBEF", "#6BAED6", "#2171B5"]
# )

# fig, ax = plt.subplots(
#     figsize=(0.38 * em_plot.shape[1] + 2.6,
#              0.33 * em_plot.shape[0] + 1.6)
# )

fig, ax = plt.subplots(
    figsize=(0.3 * em_plot.shape[1] + 1.6,
             0.3 * em_plot.shape[0] + 1.6)
)


ax = sns.heatmap(
    em_plot,
    cmap=chromhmm_white_purple,
    # cmap="Blues",
    vmin=0,
    vmax=100,
    annot=True,
    fmt=".0f",
    linewidths=0.0,
    linecolor="white",
    cbar=False
)

# Make all annotation text white
for text in ax.texts:
    text.set_color("white")
    
        
# Put marks on top (like ChromHMM)
ax.xaxis.tick_top()
ax.xaxis.set_label_position('top')

ax.set_xticklabels(ax.get_xticklabels(), rotation=90)
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title("ChromHMM Emissions", pad=20)

plt.tight_layout()
fig.savefig("Histone_chromhmm_emissions.pdf", bbox_inches="tight")   # vector
plt.show()
```
