# Public-source note

Intersects K=4 RNA-cluster TSS ±1 kb intervals with starting-cell H3K27ac/H3K4me3 peaks and performs histone-mark filtering. Outputs were stripped and paths sanitized.

```python
import os, shlex, subprocess, sys
import pandas as pd

import os, shlex, subprocess, sys
import pandas as pd
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import LinearSegmentedColormap
```

```python
histone_peak_dir="/path/to/CEBPB_seq/Chip/histone_all_samples_dedup_macs2"
RNA_cluster_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/"
H3K27ac_out_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/intersectBed_H3K27ac/"
H3K4me3_out_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/intersectBed_H3K4me3/"


H3K27ac_filter_by_H3K4me3_out_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/intersectBed_H3K27ac/filter_by_H3K4me3"
H3K4me3_filter_by_H3K27ac_out_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/intersectBed_H3K4me3/filter_by_H3K27ac/"

H3K27ac_filter_by_other_H3K27ac_out_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/intersectBed_H3K27ac/filter_by_other_H3K27ac/"


filter_by_other_H3K4me3_out_dir="/path/to/CEBPB_seq/output/bulk_rna/Cluster_all_starting_cell_types/Starting_cell_types_k-means_clusters_k4_tss_1kb_beds/intersectBed_H3K4me3/filter_by_H3K27ac/filter_by_other_H3K4me3/"


K_clusters =4

```

```python
import glob
keyword = "NSC"
glob.glob(os.path.join(histone_peak_dir, f"{keyword}*ummits.bed"))
```

## intersectBed with clusters TSS +- 1kb H3K27ac

```python
H3K27ac_peak_summits_dict={    'MEF':'MEF_FBS_Cell2017_H3K27ac_mm10_sorted_dedup_macs2_summits.bed',
       'NSC':'NSC_H3K27AC_mm10_sorted_dedup_macs2_summits.bed',
    'ESC':'ESC_H3K27ac_mm10_sorted_dedup_macs2_summits.bed',
        'Bcell':'BMpreB_H3K27Ac_ChIP_rep1_mm10_sorted_dedup_macs2_summits.bed'

    
}
len(H3K27ac_peak_summits_dict.keys()),len(H3K27ac_peak_summits_dict.values())
```

```python
mark="H3K27ac"
for cl in range(1,K_clusters+1):
    print(cl)
    cl_bed_path= RNA_cluster_dir+"/RNA_cluster_"+str(cl)+".bed"
    print(cl_bed_path)
    for key, value in H3K27ac_peak_summits_dict.items():
        print(key, value)
        peak_summits_path=histone_peak_dir+'/'+value
        output_bed=H3K27ac_out_dir+"/RNA_cluster_"+ str(cl)+"_intersectBed_"+key+"_"+mark+".bed"
        print(output_bed)
        cmd = [
        "intersectBed",
        "-a", str(peak_summits_path),
        "-b", str(cl_bed_path),
        # "-wa"  # write entries from A that overlap B
        "-wb",  # uncomment if you also want entries from B
        ]
    
        with open(output_bed, "w") as out:
            subprocess.run(cmd, stdout=out, check=True)
            print(f"Intersection written to {output_bed}")
    #     break
    # break
```

```python
for key, value in Cluster_label_dict.items():
    print(key, value)
    filtered_fpth=H3K27ac_filter_by_H3K4me3_out_dir+"/"+value+"_intersectBed_"+key+"_"+"H3K27ac"+".bed"
    H3K4me3_file=H3K4me3_out_dir+"/"+value+"_intersectBed_"+key+"_"+"H3K4me3"+".bed"
    H3K27ac_file=H3K27ac_out_dir+"/"+value+"_intersectBed_"+key+"_"+"H3K27ac"+".bed"
    # print(H3K27ac_file)
    # print(H3K4me3_file)
    H3K27ac_df=pd.read_csv(H3K27ac_file, header=None, sep='\t')
    H3K4me3_df=pd.read_csv(H3K4me3_file, header=None, sep='\t')
    print("len(H3K27ac_df)",len(H3K27ac_df))
    H3K27ac_sub_df=H3K27ac_df[H3K27ac_df[8].isin(set(H3K4me3_df[8]))]

    H3K27ac_sub_df.to_csv(filtered_fpth, index=None, header=None, sep='\t')
    print("len(H3K27ac_sub_df)",len(H3K27ac_sub_df))
    print("len(set(H3K27ac_sub_df[8]))",len(set(H3K27ac_sub_df[8])))
    # break;
```
